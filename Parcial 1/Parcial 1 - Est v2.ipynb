## Ejercicio 1: Estadísticas sin Fondo con NumPy Vectorizado (Clase 1)
**Contexto:** En un pipeline de microscopía, las imágenes en escala de grises llegan como matrices de NumPy donde el fondo es negro (valor 0) y las células son brillantes. Necesitas un módulo reutilizable que calcule estadísticas solo sobre las células.

**Instrucción:** Crea una clase `ImageStats` que reciba una matriz de NumPy (imagen en escala de grises) y tenga dos métodos **sin usar bucles `for` explícitos sobre píxeles**:
1. `normalizar()`: retorna la imagen reescalada al rango $[0, 1]$ (normalización min-max).
2. `estadisticas()`: retorna la media y la desviación estándar **ignorando los píxeles con valor 0** (fondo), usando indexación booleana.

**Pregunta para el examen:**
> ¿Por qué es estadísticamente relevante excluir el fondo (valor 0) al calcular la media en este tipo de imágenes? ¿Y por qué la indexación booleana es preferible a recorrer la matriz con bucles `for` anidados?

Agrega títulos a cada subplot.

## Ejercicio 3: Segmentación Bicolor en HSV (Clase 3)
**Contexto:** En control de calidad debes separar piezas rosadas y verdes sobre una banda transportadora. Trabaja con la imagen https://raw.githubusercontent.com/UN-GCPDS/ProcesamientoDigitalImagenes/refs/heads/main/Parcial%201/Figuras/circles.png .

**Instrucción:** Escribe un script que: (1) cargue la imagen con `cv2.imread` (recuerda que OpenCV usa orden BGR), (2) convierta a HSV con `cv2.cvtColor`, (3) defina **dos rangos** con `cv2.inRange` (uno para tonos rosas/magenta y otro para verdes), (4) combine ambas máscaras con OR lógico (`cv2.bitwise_or`) y (5) aplique la máscara a la imagen con `cv2.bitwise_and`.

**Pregunta para el examen:**
> ¿Por qué es preferible segmentar por color en el espacio HSV en lugar de RGB? ¿Por qué en OpenCV el canal H va de 0 a 179 y no de 0 a 360?

## Ejercicio 4: Limpieza Morfológica de Binaria (Clase 4)
**Contexto:** Tras binarizar el escaneo de un documento, obtienes ruido de sal (puntos blancos aislados sobre el fondo) y pequeños huecos negros dentro de los caracteres.

**Instrucción:** Implementa una función `limpiar(binaria)` que aplique, en orden: (1) una **apertura** con elemento estructurante elíptico de $3 \times 3$ (`cv2.getStructuringElement` + `cv2.morphologyEx` con `MORPH_OPEN`) para eliminar el ruido de sal, y (2) un **cierre** con elemento elíptico de $9 \times 9$ (`MORPH_CLOSE`) para tapar los huecos. No uses implementaciones manuales de erosión/dilatación.

**Pregunta para el examen:**
> ¿Cuál es el riesgo de usar un elemento estructurante demasiado grande en la apertura cuando hay detalles finos? ¿Qué pasaría si inviertes el orden (cierre primero, apertura después) en una imagen con ruido de sal?

## Ejercicio 5: Conteo con Umbral Adaptativo, Canny y Contornos (Clases 3 y 4)
**Contexto:** Debes contar objetos circulares (monedas o células) en fotos con iluminación irregular: un umbral global falla en las zonas de sombra.

**Instrucción:** Implementa un script que: (1) binarice con **umbral adaptativo gaussiano** (`cv2.adaptiveThreshold` con `ADAPTIVE_THRESH_GAUSSIAN_C`), (2) extraiga bordes de la imagen en grises con **Canny** (`cv2.Canny`), y (3) encuentre contornos externos (`cv2.findContours` con `RETR_EXTERNAL`) y **cuente solo los de área mayor a un umbral** (`cv2.contourArea`) para descartar ruido.

**Pregunta para el examen:**
> El algoritmo de Canny incluye la "supresión de no-máximos" y la "histéresis con dos umbrales". ¿Qué hace cada paso y qué le pasaría a los bordes si se omitieran?
